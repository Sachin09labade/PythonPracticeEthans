```python
#slicing the list

List_01=["apple","banana","mango","pineapple"]
# +ve index   0      1       2         3
# -ve index  -4     -3      -2        -1

print(List_01[1:3])
print(List_01[-3])
```

```python
#append the list
List_01=["apple","banana","mango","pineapple"]
List_01.append("kiwi")
List_01
```

```python
#insert into the list
List_01=["apple","banana","mango","pineapple"]
List_01.insert(2,"kiwi")
List_01
```

```python
#add multiple items 
List_01=["apple","banana","mango","pineapple"]
List_02=["banana","mango","pineapple"]
L3=List_01+List_02
L3

```

```python
#remove specified item
List_01=["apple","banana","mango","pineapple"]
List_01.remove("banana")
List_01
```

```python
#remove by index
List_01=["apple","banana","mango","pineapple"]
List_01.pop(2)
List_01
```

```python
#delete by index
List_01=["apple","banana","mango","pineapple"]
del List_01[1]
List_01
```

```python
#remove last item
List_01=["apple","banana","mango","pineapple"]
List_01.pop()
List_01
```

```python
#clear all items
List_01=["apple","banana","mango","pineapple"]
List_01.clear()
List_01
```

```python
#delete entire list
List_01=["apple","banana","mango","pineapple"]
del List_01
List_01
```

```python
#assignment/reference copy
l1=["1","2","3"]
l2=l1         #both l1 and l2 point to the same memory location
l2
```

```python
#shallow copy
l1=["1","2","3"]
l2=l1.copy()
l2
```

```python
#joining/merging
l1=["1","2","3"]
l2=["2","5","7"]
l3=l1+l2
l3
```

```python
#extending the list
l1=["1","2","3"]
l2=["2","5","7"]
l1.extend(l2)
l1
```

```python
#sorting the list
l1=["4","2","3","7"]
l1.sort()
l1
```

```python
#sorted copy of the list
sorted(l1)  #sorting list in new memory location
```

```python
#sort reverse
l1=["1","2","3"]
l1.sort(reverse=True)
l1
```

```python
#reversing the list
l1=["1","2","3","4"]
l1.reverse()
l1
```

```python
#reverse using slicing
l1=["1","2","3","4"]
l1=l1[::-1]
l1
```

```python
#search index of the item
l1=["1","2","3","4"]
l1.index("3")
```

```python
#count items in the list
l1=["1","2","3","4","3","3"]
l1.count("3")
```

```python
#length of the list
l1=["1","2","3","4","5","6"]
len(l1)
```

```python
#create the list from the existing list where you need to add the square of the list_01 items
list_01= [1,2,3,4]
newList = []
for i in list_01:
    newList.append(i*i)
newList
```
