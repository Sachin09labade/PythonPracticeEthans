```python
#create dict
myDict={"Name":"Rajesh","Age":30}
myDict
```

```python
#by using constructor
myDict=dict(Name="Rajesh", Age=30)
myDict
```

```python
#create dict with list as value
myDict={"Name":["Sunil","Rajesh","Aniket"],"Age":[30, 25, 20],"Education":["BA","Bsc","Bcom"]}
myDict
```

```python
#accessing values from dict
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict["name"]
```

```python
#accessing values from dict using get() method
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict.get("age")
```

```python
#accessing only keys from dict
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict.keys()
```

```python
#accessing only values from dict
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict.values()
```

```python
#accessing both keys and values from dict
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict.items()
```

```python
#iterating through dict keys
myDict = {"name": "John", "age": 25, "city": "Pune"}
for i in myDict.keys():
    print(i)
```

```python
#iterating through dict values
myDict = {"name": "John", "age": 25, "city": "Pune"}
for i in myDict.keys():
    print(myDict[i])
```

```python
#iterating through dict values using get() method
myDict = {"name": "John", "age": 25, "city": "Pune"}
for i in myDict.keys():
    print(myDict.get(i))
```

```python
#iterating through dict values using values() method
myDict = {"name": "John", "age": 25, "city": "Pune"}
for i in myDict.values():
    print(i)
```

```python
#iterating through dict items
myDict = {"name": "John", "age": 25, "city": "Pune"}
for i,j in myDict.items():
    print(i,j)
```

```python
#pop() method removes the item with specified key name
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict.pop("city")
myDict
```

```python
#popitem() method removes the last item from dict
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict.popitem()
myDict
```

```python
#del keyword removes item with specified key name
myDict = {"name": "John", "age": 25, "city": "Pune"}
del myDict["age"]
myDict
```

```python
#clear() method removes all items from dict
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict.clear()
myDict
```

```python
#delete entire dictionary
myDict = {"name": "John", "age": 25, "city": "Pune"}
del myDict
myDict
```

```python
#Adds a new key-value pair to the dictionary
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict["country"] = "India"
myDict
```

```python
#Changes the value if the key already exists
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict["age"] = 30
myDict
```

```python
# Adds new items or updates existing key-value pairs
myDict = {"name": "John", "age": 25, "city": "Pune"}
myDict.update({"age": 30, "country": "India",})
myDict
```

```python
#Creating a copy of the dictionary
myDict = {"name": "John", "age": 25, "city": "Pune"}
newDict = myDict.copy()
newDict
```

```python
#Creating a reference of the dictionary (both variables point to same dictionary in memory location)
myDict = {"name": "John", "age": 25, "city": "Pune"}
newDict=myDict
newDict
```

```python
#length of dictionary
myDict = {"name": "John", "age": 25, "city": "Pune"}
len(myDict)
```

```python
#checking if a key exists in dictionary using IN operator
myDict = {"name": "John", "age": 25, "city": "Pune"}
"name" in myDict
```

```python
#reverse the dictionary
myDict = {"name": "John", "age": 25, "city": "Pune"}
dict(reversed(list(myDict.items())))
```

```python
#merge two dictionary
Dict_01={"Name":"Rajesh","Age":30}
Dict_02={"Pin":444001,"Area":"Sanghavi"}
newDict={**Dict_01,**Dict_02}
newDict
```

```python
#merge two dictionary using dict() constructor
Dict_01={"Name":"Rajesh","Age":30}
Dict_02={"Pin":444001,"Area":"Sanghavi"}
newDict=dict(Dict_01,**Dict_02)
newDict
```

```python
#sorting the dictionary
myDict = {"name": "John", "age": 25, "city": "Pune"}
newDict = dict(sorted(myDict.items()))
newDict
```

```python
#Join the two dictionaries using update() method
Dict_01 = {"name":"Rajesh","Age":24}
Dict_02 = {"Pin":444001,"Area":"Sanghavi"}
Dict_01.update(Dict_02)
Dict_01
```

```python
#square of numbers using dictionary comprehension
list_01 =  [1,2,3,4,5]
NewDict_01 = {}
for i in list_01:
    NewDict_01[i] = i*i
NewDict_01
```

```python
#dictionary comprehension
list_01 =  [1,2,3,4,5]
{i: i*i for i in list_01}
```
