```python
#access tuple elements
myTuple = ("apple", "banana", "mango")
myTuple[1]   #+ve indexing

myTuple[-1]   #-ve indexing
```

```python
#slicing in tuple
myTuple = ("apple", "banana", "mango")
myTuple[1:3]   
```

```python
#returns the tuple in reverse order (storing in new memory location)
myTuple = ("apple", "banana", "mango")
myTuple[::-1]
```

```python
#search element by index
myTuple = ("apple", "banana", "mango")
myTuple.index("mango")
```

```python
#count the number of times an element is repeated in tuple
myTuple = ("apple", "banana", "mango","apple","kiwi","apple")
myTuple.count("apple")
```

```python
#joining two tuples
myTuple1 = ("apple", "banana", "mango")
myTuple2 = ("apple", "banana", "mango")
tup3=myTuple1+myTuple2
tup3
```

```python
#IN operator in tuple
myTuple1 = ("apple", "banana", "mango")
"apple" in myTuple1
```

```python
#adding an element to a tuple
myTuple = ("apple", "banana")
myTuple += ("mango",)
myTuple
```

```python
#Repeats the tuple the specified number of times.
myTuple = ("apple", "banana", "mango")
myTuple * 2
```

```python
#checking the length of tuple
myTuple = ("apple", "banana", "mango")
len(myTuple)
```

```python
#assigning a tuple to another tuple
myTuple = ("apple", "banana", "mango")
tuple2=myTuple
tuple2
```

```python
#converting tuple to list
myTuple = ("apple", "banana", "mango")
list(myTuple)
```

```python
#converting list to tuple
myList= ["apple", "banana", "mango"]
tuple(myList)
```
