```python
#adding element in set
mySet = {"apple", "banana", "mango"}
mySet.add("orange")
mySet
```

```python
#update (adding multiple elements in set)
mySet = {"apple", "banana", "mango"}                             #abc=["orange","grapes"]
mySet.update(["orange","grapes"])                                #mySet.update(abc)     #both are same
mySet
```

```python
#remove an item
mySet = {"apple", "banana", "mango"}
mySet.remove("banana")
mySet
```

```python
#removes(discards) the item and does not give an error if the item does not exist
mySet = {"apple", "banana", "mango"}
mySet.discard("banana")
mySet
```

```python
#removes(pop) an arbitrary item from the set
mySet = {"apple", "banana","grapes"}
mySet.pop()
mySet
```

```python
#clears the set
mySet = {"apple", "banana","grapes"}
mySet.clear()
mySet
```

```python
#delete the set
mySet = {"apple", "banana","grapes"}
del mySet
mySet
```

```python
#shallow copy of the set
mySet = {"apple", "banana","grapes"}
newSet=mySet.copy()
newSet
```

```python
#assignment/reference copy
mySet = {"apple", "banana","grapes"}
newSet=mySet    #both mySet and newSet point to the same memory location
newSet
```

```python
#union of two sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set3 = set1.union(set2)
set3
```

```python
#intersection of two sets
set1 = {1, 2, 3}
set2 = {2, 3, 4}
set3 = set1.intersection(set2)
set3
```

```python
#difference of two sets
set1 = {1, 2, 3}
set2 = {2, 3, 4}
set3 = set1.difference(set2)
set3
```

```python
#symmetric difference of two sets
set1 = {1, 2, 3}
set2 = {2, 3, 4}
set3 = set1.symmetric_difference(set2)
set3
```
