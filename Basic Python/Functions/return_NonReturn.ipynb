```python
#return type function
def add_values(x,y):
    z=x+y
    return z
values=add_values(5,6)
values

def newValue(c,b):
    t=c+b
    return t
values_01=newValue(values,7)
print(values_01)
```

```python
#non return type function
def add_values(x,y):
    z=x+y
    print(z)
values=add_values(5,6)
print(values)    
```

```python
#use of pass keyword (empty function)
def my_function():
    pass
my_function()
```

```python
#use of yield keyword
def newFunc(a):
    for i in range(1,a):
        yield i
values_1=newFunc(10)

print(next(values_1))
print(next(values_1))
print(next(values_1))
```

```python
#using return keyword in the position of yeild keyword
def newFunc_01(a):
    for i in range(1,a):
        return i
values=newFunc_01(10)
print(values)
print(values)
print(values)
```
