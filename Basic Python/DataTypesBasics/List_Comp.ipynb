```python
#printing numbers from 1 to 10
for i in range(1,10):
    print(i)
    
```

```python
#printing numbers from 1 to 10 using list comprehension
[i  for i in range(1,10)]
```

```python
#[exp Loop if_condition]
#[exp for i in iterable if_condition]
#print even numbers using list comprehension
[i  for i in range(1,10)  if i%2==0]
```

```python
#print square of even numbers using list comprehension
[i**2  for i in range(1,10)  if i%2==0]
```

```python
#printing numbers less than 2 and greater than 5 and then storing it in list 
newList=[]
for i in range(0,10):
    if i<2 or i>5:
        newList.append(i)
newList        
```

```python
#printing numbers less than 2 and greater than 5 using list comprehension
[i  for i in range(0,10)  if i<2 or i>5]          #using multiple conditions    #used OR operator
```

```python
#printing even numbers that are greater than 5 and then storing it in list
newList=[]
for i in range(0,10):
    if i>5 and i%2==0:
        newList.append(i)
newList   
```

```python
#printing even numbers that are greater than 5 using list comprehension 
[i  for i in range(0,10)  if i%2==0 and i>5]        #using multiple conditions    #used AND operator
```

```python
#using for loop and multiple if-else conditions and then storing it in list
newList=[]
for i in range(0,15):
    if i<5:
        newList.append(i*2)
    elif i<10:
        newList.append(i/2)
    else:
        newList.append(i+2)
newList                
```

```python
#using for loop and multiple if-else conditions in list comprehension
[i*2   if i<5   else i/2   if i<10   else i+2   for i in range(0,15)]
```

```python
#convert below in dict comprehension
#[i*2   if i<5   else i/2   if i<10   else i+2   for i in range(0,15)]
{i:   i*2   if i<5   else i/2   if i<10   else i+2   for i in range(0,15)}
```
