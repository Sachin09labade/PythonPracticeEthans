```python
#ZeroDivisionError in exception handling
a=5
b=0
try:
    a/b
except Exception:
    print("We can not divide a value by zero")    
```

```python
#try, except and finally block
a=5
b=2
try:
    c=a/b
    print(c)
except:
    print("Something went wrong")
finally:
    print("Program closed")
```

```python
#try, except and else block
a=5
b=2
try:
    c=a/b
    print(c)
except:
    print("Something went wrong")
else:
    print("Program without any error")
```
