```python
#if, elif and else
i=3
if i<5:
    print("Value is less than 5")
elif i<7:
    print("Value is less than 7")
elif i<10:
    print("Value is less than 10")
else:
    print("Value is greater")
```

```python
#using lists in if,elif and else
Stall_1=["Apple","Mango","Pineapple","Watermelon"]
Stall_1_price=150

Stall_2=["Mango","Pineapple","Watermelon","Apple"]
Stall_2_price=180

if "Apple" in Stall_1 and Stall_1_price<=150:
    print("Buy apple from Stall_1")
elif "Apple" in Stall_2 and Stall_2_price<=150:
    print("Buy apple from Stall_2")
else:
    print("Apples are not within the price range")



```
