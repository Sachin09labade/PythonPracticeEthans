```python
#creating a file and writing into it
FilePath="file_02.txt"
file=open(FilePath,"w")
file.write("This is a python class")
file.close()
```

```python
#reading data from a created file
FilePath="file_02.txt"
file=open(FilePath,"r")
file.readline()
```

```python
#appending data on a created file
FilePath="file_02.txt"
file=open(FilePath,"a")
file.write(" Shivajinagar Branch")
file.close()
```

```python
#reading data from a file and printing it in newlines by using for loop
FilePath = "file_02.txt"
file = open(FilePath,"r")
newLines = file.readlines()

for i in newLines:
    print(i)
```

```python
#deleting or removing a file 
import os
FilePath = "file_01.txt"      #here we have deleted or removed the file_01.txt file created at the start
file = open(FilePath,"r")
file.close()
os.remove(FilePath)
```

```python
#creating a new file, writing into it, reading data from it through with keyword
FilePath="file_03.txt"
with open(FilePath,"w") as file:                   #with keyword used to close the file automatically
    file.write("Today is 14th october")

with open(FilePath,'r') as file:                   #with keyword used to close the file automatically
    print(file.read())    
```
