```python
#class and object using if-else and for loop
class test:
    
    def abc(self,i):
        if i<10:
            print('Less than 10')
        else:
            print("Greater than 10")
    def xyz(self,i):
        for i in range(0,i):
            print(i)

newObj = test()
newObj.abc(2)     #passing value when calling the function
newObj.xyz(5)     #passing value when calling the function
```

```python
#class and object program using __init__ function
class test:
    def __init__(self,someIntValue):
        self.someIntValue = someIntValue
    def abc(self):
        if self.someIntValue<10:
            print('Less than 10')
        else:
            print("Greater than 10")
    def xyz(self):
            for x in range(0,self.someIntValue):
                print(x)
  
newObj = test(someIntValue=11)    #giving the value to all the functions through class
newObj.abc()
newObj.xyz()
```
