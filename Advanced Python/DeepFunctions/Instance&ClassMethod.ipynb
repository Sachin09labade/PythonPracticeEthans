```python
#program to print age of dog
class Dog():
    def __init__(self):
        self.dog_Age=14         #instance variable

dog_1=Dog()
dog_1.dog_Age     
```

```python
#program to print age and color of dog
class Dog():
    def __init__(self,dog_Age):
        self.dog_Age=dog_Age           #instance variable

    def abc(self):                     #instance method    
        print(self.dog_Age)    

    def my_dog_details(self,colour):                 #instance method with argument
        print(f"My dog has {colour} color")           #f is used for formatted string

dog_1=Dog(14)
dog_1.abc() 
dog_1.my_dog_details("Brown")
```

```python
class Banking:
    Int_Rate=0.08     #8% interest rate  #class variable

    @classmethod                     #class method
    def Change_Int_Rate(cls): 
        cls.Int_Rate=cls.Int_Rate+(cls.Int_Rate*0.02)
        print(cls.Int_Rate)

Banking.Int_Rate
Banking.Change_Int_Rate()
```
