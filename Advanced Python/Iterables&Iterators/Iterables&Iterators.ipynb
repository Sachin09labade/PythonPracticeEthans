```python
#example of iterable
friutsList=["apple","banana","cherry","orange"]
print(dir(friutsList))

for i in friutsList:
    print(i)
```

```python
#example of iterator
'''Apply Filter'''
friutsList=[1,2,3,4,5,6,7,8]
def extractOnlyEvenNumbers(i):
    return i%2==0
abc=filter(extractOnlyEvenNumbers,friutsList)           #using filter function
print(abc)
print(dir(abc))

for i in abc:
    print(i)
    
```

```python
#example to create an iterator from iterable
mystr="banana"
myit=iter(mystr)
print(myit)
print(dir(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```

```python
#example to create an iterator from iterable
fruit_tuple=("apple","banana","cherry")
myit=iter(fruit_tuple)
print(myit)
print(dir(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```

```python
#printing 1 to 5 numbers from list using iterator
list_01=[1,2,3,4,5]
new_Iterator=iter(list_01)

print(next(new_Iterator))
print(next(new_Iterator))
print(next(new_Iterator))
print(next(new_Iterator))
print(next(new_Iterator))
```

```python
#printing 1 to 5 numbers using iterator in for loop
list_01 = [1,2,3,4,5]
new_Iterator = iter(list_01)

for i in new_Iterator:
    print(i)
```

```python
#printing even numbers using iterator and stored in list
newRange=range(1,10)
def new_Func(i):
    if i%2==0:
        return i
new_iterator=filter(new_Func,newRange)
print(list(new_iterator))    
```

```python
#Enumerate
#example of enumerate
Fruit_List=["apple","cherry","banana"]
enu_fruits=enumerate(Fruit_List)
enu_fruits

for index,value in enu_fruits:            #we can also use i,j instead of index,value
    print(index,value)
```
