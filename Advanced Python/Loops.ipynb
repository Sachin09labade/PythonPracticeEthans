```python
#printing numbers using for loop
for j in range(1,7):
    print(j,end=" ")
```

```python
#printing number pattern in square
for i in range(1,7):
    for j in range(1,7):
        print(j,end=" ")
    print()
```

```python
#printing hash pattern in square
for i in range(1,7):
    for j in range(1,7):
        print("#",end=" ")
    print()
```

```python
#printing star pattern in square
for i in range(1,7):
    for j in range(1,7):
        print("*",end=" ")
    print()
```

```python
#printing half pyramid of numbers
for i in range(1,7):
    for j in range(1,i):
        print(j,end=" ")
    print()
```

```python
#printing half pyramid of stars
for i in range(1,7):
    for j in range(1,i):
        print("*",end=" ")
    print()
```

```python
#printing half pyramid of hashes
for i in range(1,7):
    for j in range(1,i):
        print("#",end=" ")
    print()
```

```python
#using nested loop to print two lists elements
col = ["Red", "Yellow", "LightGreen"]
fruits = ["Apple", "Banana", "Grapes"]

for i in col:
    for j in fruits:
        print(i,j)
```

```python
#printing matched elements from two lists
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black","White" ,"LightGreen"]
for i in col_01:
    for j in col_02:
        if i==j:
            print("Matched",i,j)
```

```python
#printing matched elements from two lists in tuple by using function 
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black","White" ,"LightGreen"]

def func_01(col_01,col_02):
    for i in col_01:
        for j in col_02:
            if i==j:
                return "Matched",i,j

values = func_01(col_01,col_02)
values
```
