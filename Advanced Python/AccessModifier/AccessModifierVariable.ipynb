```python
#simple program of access modifiers
class Student:
    def __init__(self):
        self.name="Rahul"        #public
        self.age=20              #protected
        self.__marks=90          #private
        
    def display(self):
        print(self.name)         #public
        print(self.age)          #protected
        print(self.__marks)      #private

obj=Student()
obj.display()          
```

```python
#outside the class
class Student:
    def __init__(self):
        self.name="Rahul"        #public
        self.age=20              #protected
        self.__marks=90          #private

obj=Student()

#access outside the class
print(obj.name)         #public = allowed
print(obj.age)          #protected = allowed
print(obj.__marks)      #private = error
```

```python
#inside child class inheritance
class ParentClass:                   #parent class 
    def __init__(self):
        self.name = "Rahul"          #public
        self._age = 20               #protected
        self.__marks = 90            #private

class ChildClass(ParentClass):       #child class
    def display(self):
        print(self.name)             #public = allowed
        print(self._age)             #protected = allowed
        print(self.__marks)          #private = error

child_Obj = ChildClass()
child_Obj.display()
```

```python
#accessing student details
class Student:
    def __init__(self):
        self.name = "Rahul"    #public
        self._age = 20         #protected
        self.__marks = 90      #private

class Teacher:
    def display(self):
        student_Obj = Student() 
        print(student_Obj.name)
        print(student_Obj._age)
        print(student_Obj.__marks)

tech_Obj = Teacher()
tech_Obj.display()
```
