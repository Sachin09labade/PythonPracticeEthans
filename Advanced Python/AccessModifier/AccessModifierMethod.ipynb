```python
#Program to demonstrate accessing public, protected, and private methods within the same class
class Student:
    def display_public(self):
        print("Public Method")
    def _display_protected(self):
        print("Protected Method")
    def __display_private(self):
        print("Private Method")

    def show(self):
        self.display_public()              #allowed
        self._display_protected()          #allowed
        self.__display_private()           #allowed

obj=Student()
obj.show()
```

```python
#Program to demonstrate accessing public, protected, and private methods outside a class using an object
class Student:
    def display_public(self):
        print("Public Method")
    def _display_protected(self):
        print("Protected Method")
    def __display_private(self):
        print("Private Method")

obj=Student()
obj.display_public()           #allowed
obj._display_protected()       #allowed
obj.__display_private()        #error
```

```python
#Program to demonstrate accessing public, protected, and private methods in a child class through inheritance
class Student:
    def display_public(self):
        print("Public Method")
    def _display_protected(self):
        print("Protected Method")
    def __display_private(self):
        print("Private Method")

class Teacher(Student):
    def show(self):
        self.display_public()          #allowed
        self._display_protected()      #allowed
        self.__display_private()       #error

obj = Teacher()
obj.show()
```

```python
#Program to demonstrate accessing public, protected, and private methods of one class through an object created inside another class
class Student:
    def display_public(self):
        print("Public Method")
    def _display_protected(self):
        print("Protected Method")
    def __display_private(self):
        print("Private Method")

class Teacher(Student):
    def show(self):
        obj=Student()
        obj.display_public()            #allowed
        obj._display_protected()        #allowed
        obj.__display_private()         #error

t = Teacher()
t.show()               
      
```
