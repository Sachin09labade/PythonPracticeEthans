```python
#example of encapsulation
class Student:
    def __init__(self):        #Constructor to initialize marks
        self.__marks = 85      #private variable
    def show_marks(self):      #Method to access private variable
        print(self.__marks)    #Marks are hidden and can be accessed only through a method

s = Student()
s.show_marks()
```

```python
#another example of encapsulation
class Mobile:
    def __init__(self):       #Constructor to initialize pin
        self.__pin = 1234     #private variable
    def show_pin(self):       #Method to access private variable
        print(self.__pin)     #Mobile pin is hidden inside the phone and is accessed through a method

m = Mobile()
m.show_pin()
```
