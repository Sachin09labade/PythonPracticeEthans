```python
#Single Inheritance
class Parent:
    def ParentClassMethod(self):
        return "Parent class method: I have 50 acres of land"
class Child(Parent):
    def ChildClassMethod(self):
        return "Child class method: I have a bike"    

parent_obj=Parent()
print(parent_obj.ParentClassMethod())  
print("")

child_obj=Child()
print(child_obj.ParentClassMethod())
print(child_obj.ChildClassMethod())
      
```

```python
#Multi-level Inheritance
class GrandParent:
    def GrandParentClassMethod(self):
        return "GrandParent class method: I have 500 acres of land"
class Parent(GrandParent):
    def ParentClassMethod(self):
        return "Parent class method: I have 50 acres of land"   
class Child(Parent):
    def ChildClassMethod(self):
        return "Child class method: I have a bike"        

grandparent_obj=GrandParent()
print(grandparent_obj.GrandParentClassMethod())  
print("")

parent_obj=Parent()
print(parent_obj.GrandParentClassMethod()) 
print(parent_obj.ParentClassMethod())  
print("")

child_obj=Child()
print(child_obj.GrandParentClassMethod())
print(child_obj.ParentClassMethod())
print(child_obj.ChildClassMethod())
```

```python
#Multiple Inheritance
class Mother:
    def MotherClassMethod(self):
        return "I have gold"
class Father:
    def FatherClassMethod(self):
        return "I have house and money"
class Child(Mother,Father):
    def ChildClassMethod(self):
        return "I have books"

child_obj=Child()
print(child_obj.MotherClassMethod())
print(child_obj.FatherClassMethod())   
print(child_obj.ChildClassMethod())
```

```python
#Example of Multiple Inheritance in Different Case
class Mother:
    def MotherClassMethod(self):
        return "I have gold"
class Son(Mother):
    def SonClassMethod(self):
        return "I have PS5"
class Daughter(Mother):
    def DaughterClassMethod(self):
        return "I have books"


son_obj=Son()
print(son_obj.SonClassMethod())
print(son_obj.MotherClassMethod())
print("")

daughter_obj=Daughter()
print(daughter_obj.DaughterClassMethod())
print(daughter_obj.MotherClassMethod())
```

```python
#Homework Example of Single Inheritance
class BankAccount:
    def AccountDetails(self):
        return "Bank Account: Can deposit and withdraw money"
class SavingsAccount(BankAccount):
    def SavingsDetails(self):
        return "Savings Account: Earns interest on balance"
    
bank_obj = BankAccount()
print(bank_obj.AccountDetails())
print("")

savings_obj = SavingsAccount()
print(savings_obj.AccountDetails())
print(savings_obj.SavingsDetails())
```

```python
#Homework Example of Multi-level Inheritance
class Company:
    def CompanyPolicy(self):
        return "Company: All employees must work 8 hours daily"
class Employee(Company):
    def EmployeeWork(self):
        return "Employee: Completes assigned tasks"
class Manager(Employee):
    def ManageTeam(self):
        return "Manager: Supervises the team"

company_obj = Company()
print(company_obj.CompanyPolicy())
print("")

employee_obj = Employee()
print(employee_obj.CompanyPolicy())   
print(employee_obj.EmployeeWork())
print("")

manager_obj = Manager()
print(manager_obj.CompanyPolicy())    
print(manager_obj.EmployeeWork())     
print(manager_obj.ManageTeam())
```

```python
#Homework Example of Multiple Inheritance
class Phone:
    def PhonePower(self):
        return "Phone: Uses battery for calling and messaging"
class Camera:
    def CameraPower(self):
        return "Camera: Uses battery for taking photos and videos"
class SmartPhone(Phone,Camera):
    def Features(self):
        return "SmartPhone: Can make calls and capture photos"

smartphone_obj = SmartPhone()
print(smartphone_obj.PhonePower())
print(smartphone_obj.CameraPower())
print(smartphone_obj.Features())
```

```python
#Homework example of single inheritance (banking)
class Bank:
    def bank_details(self):
        print("Bank: SBI")
class Customer(Bank):
    def customer_details(self):
        print("Customer: Sachin")

c_obj= Customer()
c_obj.bank_details()
c_obj.customer_details()
```

```python
#Homework example of multi-level inheritance (insurance)
class Insurance:
    def insurance_type(self):
        print("Health Insurance")
class Policy(Insurance):
    def policy_number(self):
        print("Policy Number: 12345")
class Holder(Policy):
    def holder_name(self):
        print("Holder Name: Rahul")

h_obj= Holder()
h_obj.insurance_type()
h_obj.policy_number()
h_obj.holder_name()
```

```python
#Homework example of multiple inheritance (banking and insurance)
class Bank:
    def loan(self):
        print("Bank provides loan")
class Insurance:
    def policy(self):
        print("Insurance provides policy")
class Customer(Bank, Insurance):
    def customer_name(self):
        print("Customer: Sachin")

c_obj= Customer()
c_obj.loan()
c_obj.policy()
c_obj.customer_name()
```
