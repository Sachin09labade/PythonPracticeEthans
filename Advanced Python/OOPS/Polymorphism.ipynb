```python
#simple method without overloading
class myClass:
    def sum_4(self,a,b,c,d):
        return a+b+c+d
    def sum_3(self,a,b,c):
        return a+b+c
    def sum_2(self,a,b):
        return a+b

my_obj=myClass()
my_obj.sum_4(1,2,3,4)
my_obj.sum_3(1,2,3)
my_obj.sum_2(1,2) 
```

```python
#example of method overloading
class myClass:
    def sum_Overloaded(self,a=None,b=None,c=None,d=None):
        if a!=None and b!=None and c!=None and d!=None:
            return a+b+c+d
        elif a!=None and b!=None and c!=None:
            return a+b+c
        elif a!=None and b!=None:
            return a+b
        else:
            return "Please pass at least two values for sum"        

myclass_obj=myClass()
myclass_obj.sum_Overloaded(1,2,3,4)
```

```python
#example of method overriding
class Animal:
    def sound(self):
        return "Animal makes sound"
class Dog(Animal):    
    def sound(self):
        return "Bark"
class Cat(Animal):    
    def sound(self):
        return "Meow"

Dog_obj=Dog()
Dog_obj.sound()  
  
```

```python
#another example of method overriding
class P_C_Interest:
    def Interest(self):
        return 0.011
class C_C_Home_Interest(P_C_Interest):
    def Interest(self):
        return 0.08

Home_obj=C_C_Home_Interest()
Home_obj.Interest()   
     
```

```python
#method overriding with super keyword
class Animal:
    def speak(self):
        print("The animal makes a sound")
class Dog(Animal):
    def speak(self):
        super().speak()            #call the speak method from the superclass
        print("The dog barks")

dog=Dog()
dog.speak()               
```
