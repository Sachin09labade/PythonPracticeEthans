```python
#example of abstraction
from abc import ABC, abstractmethod              #Import modules required for abstraction
class Vehicle(ABC):            #Abstract class
    @abstractmethod            #Abstract method
    def start(self):
        pass
class Car(Vehicle):                       #Child class
    def start(self):                      #Providing implementation of abstract method
        print("Car starts with a key")

c = Car()
c.start()
```

```python
#another example of abstraction
from abc import ABC, abstractmethod              #Import modules required for abstraction
class ATM(ABC):             #Abstract class
    @abstractmethod         #Abstract method
    def withdraw(self):
        pass
class BankATM(ATM):                  #Child class
    def withdraw(self):              #Providing implementation of abstract method
        print("Money withdrawn")

a = BankATM()
a.withdraw()
```
