```python
#case change

text="Hello my friends"
x=text.lower()          #lowercase
print(x)

y=text.upper()          #uppercase
print(y)

z=text.capitalize()
print(z)
```

```python
#replace

txt = "I like bananas"
x = txt.replace("bananas", "apples")
print(x)
```

```python
#join

myTuple = ("Sachin", "Raghunath", "Labade")
x = "#".join(myTuple)
print(x)
```

```python
#split
text="Welcome to the jungle"
x=text.split()
print(x)
```

```python
#count (how many words are repeated)
txt = "I love apple, apple are my favorite fruit"
x = txt.count("apple")
print(x)

```

```python
#index (checking the index number of char )
text="My name is sachin"
x=text.index("name")
print(x)
```

```python
#startswith
txt = "Hello, welcome to my world."
x = txt.startswith("Hello")
print(x)

#endswith
y=txt.endswith("world.")
print(y)
```

```python
#isnumeric
txt="12345"
x=txt.isnumeric()
print(x)
```

```python
#isalpha
txt = "Sachin"
x = txt.isalpha()
print(x)
```

```python
#isalnum
txt = "Sac12"
x = txt.isalnum()
print(x)
```

```python
#extract substring
my_text="rajesh40"
substring=my_text[0:6]
print(substring)
```

```python
#extract string using for loop
txt="sachin09labade"
new_txt=""
for i in txt:
    if i.isalpha() == True:
        new_txt = new_txt + i
        print(new_txt)
print(new_txt)     


        
```
