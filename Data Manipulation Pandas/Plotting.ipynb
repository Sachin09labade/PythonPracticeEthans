```python
#importing matplotlib and pandas
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#reading csv file into the dataframe
file_path="D:/Python/Data/Airline1.csv"
df = pd.read_csv(file_path)
df.head()
```

```python
#Bar chart of Airline vs Price
plt.bar(df['Airline'],df['Price'])
plt.show()
```

```python
#Bar chart of Airline vs Price with rotated labels
plt.bar(df['Airline'],df['Price'])
plt.xticks(rotation=45)
plt.show()
```

```python
#Bar chart of Airline vs Price with labels
plt.bar(df['Airline'],df['Price'])
plt.xticks(rotation=45)
plt.xlabel("Airline")
plt.ylabel('Price')
plt.show()
```

```python
#bar chart showing Airline names and their Prices with labels and title
plt.bar(df['Airline'],df['Price'])
plt.xticks(rotation=45)
plt.xlabel("Airline")
plt.ylabel('Price')
plt.title("Airline Vs Price")
plt.show()
```

```python
#Bar chart of Airline vs Price with labels, title, and font size
font=15
plt.bar(df['Airline'],df['Price'])
plt.xticks(rotation=45)
plt.xlabel("Airline",fontsize=font)
plt.ylabel('Price',fontsize=font)
plt.title("Airline Vs Price",fontsize=font)
plt.show()
```

```python
#Bar chart of Airline vs Price with custom size, labels, and title
font=15
plt.figure(figsize=(7,3))
plt.bar(df['Airline'],df['Price'])
plt.xticks(rotation=45)
plt.xlabel("Airline",fontsize=font)
plt.ylabel('Price',fontsize=font)
plt.title("Airline Vs Price",fontsize=font)
plt.show()
```

```python
#Seaborn bar plot of Airline vs Price
import seaborn as sns
sns.barplot(x=df['Airline'],y=df['Price'])
plt.xticks(rotation=90)
plt.show()
```

```python
#line plot of airline vs price with marker
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
plt.figure(figsize=(10, 5))
sns.lineplot(x=df['Airline'],y=df['Price'],marker='o')
plt.title("Airline vs Price")
plt.xticks(rotation=90)
plt.show()
```

```python
#countplot of source with bar label
import seaborn as sns
plt.figure(figsize=(10,4))
ax=sns.countplot(x=df['Source'],width=0.5)
ax.bar_label(ax.containers[0],fmt='%d',fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
#countplot of destination with bar label
plt.figure(figsize=(10,4))
ax=sns.countplot(x=df['Destination'],width=0.5)
ax.bar_label(ax.containers[0],fmt='%d',fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
#countplot of airline with bar label
plt.figure(figsize=(10,4))
ax=sns.countplot(x=df['Airline'],width=0.5)
ax.bar_label(ax.containers[0],fmt='%d',fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
#countplot of total stops with bar label
plt.figure(figsize=(10,4))
ax=sns.countplot(x=df['Total_Stops'],width=0.5)
ax.bar_label(ax.containers[0],fmt='%d',fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
#violin plot of airline vs price
import seaborn as sns
plt.figure(figsize=(10,4))
sns.violinplot(x=df['Airline'],y=df['Price'])
plt.xticks(rotation=90)
plt.show()
```

```python
#box plot of airline vs price
plt.figure(figsize=(17,5))
sns.boxplot(y=df['Price'],hue=df['Airline'])
plt.xticks(rotation=90)
plt.show()
```
