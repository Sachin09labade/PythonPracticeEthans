```python
#reading excel file into the dataframe and importing pandas, matplotlib and seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
file_path="D:/Python/Data/Swarn_Airline.xlsx"
df = pd.read_excel(file_path)
df.head()
```

```python
#Create a bar plot showing total ticket price for each airline, sorted by total price with different parameters
sns.set_theme(style="dark",palette="pastel")
plt.figure(figsize=(10,5)) 

Airline_Sort=df.groupby('Airline')['Price'].sum().sort_values().index

ax=sns.barplot(x=df['Airline'],y=df['Price'],errorbar=None,estimator='sum',legend=False,order=Airline_Sort,alpha=.9,gap=.1)

ax.bar_label(ax.containers[0])
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
#Create a count plot showing the number of flights for each airline, sorted by flight count with different parameters
sns.set_theme(style="dark", palette="pastel")
plt.figure(figsize=(10, 5))

Airline_Sort = df.groupby('Airline')['Price'].count().sort_values().index

ax = sns.countplot(x=df['Airline'],order=Airline_Sort,alpha=0.9,gap=0.3,legend=True)

ax.bar_label(ax.containers[0])
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
#Create a line plot showing the average ticket price for each airline with different parameters
sns.set_theme(style="darkgrid",palette='Set1')
plt.figure(figsize=(10, 5))

sns.lineplot(x=df['Airline'],y=df['Price'],errorbar=None,marker='s',estimator='mean',alpha=1,linestyle="--")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
