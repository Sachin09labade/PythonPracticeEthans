```python
#example of series (pass either list or tuple)
import pandas as pd
List_01=[1,2,3,4,5]
Series=pd.Series(List_01)
Series
```

```python
#example of series using dictionary
Data_01={"Key1":1,"Key2":2,"Key3":3}
Data_01
pd.Series(Data_01)
```

```python
#creating series using dictinary pandas (new data)
calories = {"Day1": 420, "Day2": 380, "Day3": 390}
calSeries = pd.Series(calories)
calSeries
```

```python
#example of dataframe using dictionary (2 columns)
new_dict={'Name':["Rajesh","Sunil","Prakash"],
          'Age':[20,30,40]}
df=pd.DataFrame(new_dict)
df
```

```python
#example of dataframe using dictionary (3 Columns)
new_dict={'Name':["Rajesh","Sunil","Prakash"],
          'Age':[20,30,40],
          'Education':["BBA","MCA","BTech"]}
df=pd.DataFrame(new_dict)
df
```

```python
#example of dataframe using list (2 Columns)
new_list=[["Rajesh",20],
          ["Sunil",30],
          ["Prakash",40]]
df=pd.DataFrame(new_list,columns=["Name","Age"])
df
```

```python
#example of dataframe using list (3 Columns)
new_list=[["Rajesh",20,"BBA"],
          ["Sunil",30,"MCA"],
          ["Prakash",40,"BTech"]]
df=pd.DataFrame(new_list,columns=["Name","Age","Education"])
df
```

```python
#creating dataframe from dictionary of list and array
newDict = { 'Name': ['Alice', 'Bob', 'Charlie', 'David'], 
            'Age':  [25, 30, 22, 35]}
df=pd.DataFrame(newDict)
df
```

```python
#mean of series elements (average value)
list1 = [1,2,3,4,5] 
Series = pd.Series(list1)
Series.mean()
```

```python
#median of series elements (middle value)
list1 = [1,2,3,4,5,6] 
Series = pd.Series(list1)
Series.median()
```

```python
#mode of series elements (most frequent)
list1 = [1,2,3,4,5,3] 
Series = pd.Series(list1)
Series.mode()
```
