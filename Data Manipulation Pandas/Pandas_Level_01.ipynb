```python
#reading csv file into a dataframe (new final_small_airline.csv)
import pandas as pd
df=pd.read_csv("D:/Python/Data/Final_Small_Airline.csv")
df
```

```python
#changing column names of dataframe
df.columns = ['First_name_1','Last_name_1','Age_1','City_1','Airline_1','Date_of_Journey','Arrival_Time','Full Name','Fare']
df
```

```python
#adding new column to dataframe (NewFare column) using map function
def newAirfare(x):
    return int(x)+500
df['NewFare']=df['Fare'].map(newAirfare)
df
```

```python
#increasing the fare by 10% using map function in new column (NewFare_2)
def newAirfare_2(x):
    return int(x)*1.10
df['NewFare_2']=df['Fare'].map(newAirfare_2)
df
```

```python
#reading csv file into a dataframe
df=pd.read_csv("D:/Python/Data/Final_Small_Airline.csv")
df
#increasing the value of new fare by 10% by using loc function and for loop 
for i in df.index:
    df.loc[i,'NewFare']=df.loc[i,' Fare ']*1.10
df  
```

```python
#reading csv file into a dataframe
df=pd.read_csv("D:/Python/Data/Final_Small_Airline.csv")
df
#increasing the value of new fare by 10% by using loc function and for loop (another way)
range_1=range(len(df))
for i in range_1:
    abc=df.loc[i," Fare "]*1.10
    df.loc[i,'NewFare']=int(abc)
df    
```
