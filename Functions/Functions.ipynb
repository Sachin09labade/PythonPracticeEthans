```python
#no arguments
def combineNames():
    Fullname="Rajesh"+" "+"Singh"
    print(Fullname)

combineNames()    
```

```python
#with arguments
list_01=[1,2,3]
def printSquares(list_01):
    NewDict_12=""
    NewDict_12={}
    for i in list_01:
        NewDict_12[i]=i*i
    print(NewDict_12)
printSquares(list_01)        

#new list_02
list_02=[2,3,4]
printSquares(list_02)

#new list_03
list_03=[10,20,30]
printSquares(list_03)

```

```python
#adding names by using position and keyword arguments in functions
def combineNames(Firstname,Lastname):
    Fullname=Firstname+" "+Lastname
    print(Fullname)

#With Position Arguments
combineNames("Sunil","Patil")

#with keyword arguments
combineNames(Firstname="Sunil",Lastname="Patil")
```

```python
#giving values to argument directly
Firstname = "Sunil"
LastName = "Singh"
combineNames(Firstname,LastName)
```

```python
#lambda function
newFunc=lambda a : a+10
#namefunction lambda argument:expression 
newFunc(4)
```

```python
#addition of 2 numbers using lambda
newFunc=lambda x,y : x+y
newFunc(3,4)
```

```python
#dict comprehension using function
list_01=[1,2,3]
def printSquares(list_01):

  return{i: i*i for i in list_01}
    
printSquares(list_01)     
```

```python
#update the dictionary using function
Dict_01={"name":"Rajesh","age":24}
Dict_02={"Pin":444001,"Area":"Sanghavi"}

def updatedicts(Dict_01,Dict_02):
    Dict_01.update(Dict_02)
    print(Dict_01)
updatedicts(Dict_01,Dict_02)    
```

```python
#save values in new list which are less than 5
list_02=[12,2,10,11,5,4]
def myfunction(list_02):
    newlist=[]
    for i in list_02:
        if i<5:                                            
            newlist.append(i)
    return newlist
    
myfunction(list_02)
```
